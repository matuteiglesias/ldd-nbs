# Ejercicio 26: Consultas en BigQuery (Solo usando la Interfaz Web)

En este ejercicio, los estudiantes se familiarizarán con la ventana de consulta de BigQuery en Google Cloud utilizando datasets públicos. El objetivo es que los estudiantes aprendan a ejecutar consultas SQL directamente en la consola de BigQuery, explorando y analizando datos abiertos.

## Objetivos

- Aprender a utilizar la consola de Google Cloud para ejecutar consultas en BigQuery.
- Explorar y analizar datasets públicos disponibles en Google Cloud.
- Practicar consultas SQL básicas y avanzadas en BigQuery.
- Interpretar los resultados obtenidos de las consultas.

## Entrada de Datos

- Dataset público: `bigquery-public-data.new_york_trees.tree_census_2015`
- Dataset público: `bigquery-public-data.usa_names.usa_1910_2013`

## Código Base

A continuación, se presentan ejemplos de consultas SQL que deben ser ejecutadas en la consola de BigQuery. Los estudiantes deben completar las tareas y responder las preguntas planteadas.

```sql
-- Consulta para contar el número de filas en la tabla seleccionada
SELECT COUNT(*) FROM `bigquery-public-data.new_york_trees.tree_census_2015`;

-- Consulta para obtener las 10 especies más comunes de árboles en Nueva York
SELECT spc_common, COUNT(*) AS cantidad
FROM `bigquery-public-data.new_york_trees.tree_census_2015`
GROUP BY spc_common
ORDER BY cantidad DESC
LIMIT 10;

-- Consulta para filtrar los árboles con diámetro mayor a 50 cm
SELECT COUNT(*)
FROM `bigquery-public-data.new_york_trees.tree_census_2015`
WHERE tree_dbh > 50;

-- Consulta para listar los 10 nombres más comunes en EE.UU. desde 1950
SELECT name, SUM(number) AS total_uso
FROM `bigquery-public-data.usa_names.usa_1910_2013`
WHERE year >= 1950
GROUP BY name
ORDER BY total_uso DESC
LIMIT 10;
```

## Tareas

1. Acceder a BigQuery y explorar los datasets públicos disponibles.
2. Ejecutar las consultas SQL proporcionadas en la consola de BigQuery.
3. Interpretar los resultados obtenidos de las consultas.

## Pruebas

Ejecutar las consultas en la consola de BigQuery y responder las siguientes preguntas:

- ¿Cuántos registros tiene la tabla seleccionada?
- ¿Cuáles son las especies de árboles más comunes en Nueva York?
- ¿Hay más árboles en parques o en veredas?
- ¿Cuáles son los nombres más populares en EE.UU. en los últimos 50 años?

## Resultado Esperado

Los estudiantes deben ser capaces de ejecutar consultas SQL en la ventana de BigQuery, explorar datasets públicos y analizar los resultados obtenidos. Además, deben ser capaces de interpretar los resultados y responder las preguntas planteadas.
