# Ejercicio 30: Diseño de DER para una Productora de Cine

En este ejercicio, se requiere modelar una base de datos relacional para una productora de cine. El objetivo es crear un Diagrama Entidad-Relación (DER) que represente la información sobre películas, directores, actores y productoras, así como las relaciones entre estas entidades.

## Objetivos

- Diseñar un Diagrama Entidad-Relación que modele adecuadamente las entidades y relaciones de una productora de cine.
- Identificar y documentar atributos clave para cada entidad.
- Asegurar que el modelo respete las restricciones y políticas de la productora.

## Entrada de Datos

1. Información de las películas:
   - Identificador único (código IMDB)
   - Título
   - Año de estreno
   - Lista de actores
   - Géneros (terror, acción, drama, etc.)
   - Resumen de la trama
   - Continuación de otra película (si aplica)

2. Información de los directores:
   - Nombre
   - Fecha de nacimiento
   - Datos de contacto

3. Información de los actores:
   - Nombre
   - Fecha de nacimiento
   - Papel desempeñado en cada producción

4. Información de las productoras:
   - Nombre
   - Dirección

1. Verificar que cada película tiene un único director.
